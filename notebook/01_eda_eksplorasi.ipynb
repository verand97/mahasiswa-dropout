## 📖 KAMUS DATA (TERJEMAHAN 37 ATRIBUT DALAM BAHASA SEHARI-HARI)

### A. Kelompok Profil Pribadi & Demografi
1. `Marital_status`: Status perkawinan (1=Lajang, 2=Menikah, 3=Duda/Janda, 4=Bercerai).
2. `Nacionality`: Kewarganegaraan mahasiswa (1=Warga Lokal/Portugal, lainnya kode negara asing).
3. `Displaced`: Mahasiswa rantau (1=Anak rantau/berasal dari luar kota, 0=Warga lokal dekat kampus).
4. `Gender`: Jenis kelamin (1=Laki-laki, 0=Perempuan).
5. `Age_at_enrollment`: Usia mahasiswa saat pertama kali masuk kuliah (rentang usia 17 - 70 tahun).
6. `International`: Status mahasiswa asing/internasional (1=Ya, 0=Bukan).

### B. Kelompok Jalur Masuk & Latar Belakang Akademik Awal
7. `Application_mode`: Jalur pendaftaran masuk kampus (seperti jalur reguler, prestasi, pindahan).
8. `Application_order`: Urutan pilihan jurusan saat mendaftar (pilihan ke-1, ke-2, s/d ke-6. Mahasiswa yang masuk di pilihan terakhir cenderung kurang berkomitmen).
9. `Course`: Kode program studi / jurusan yang diambil (Teknik Informatika, Manajemen, Desain, dll.).
10. `Daytime_evening_attendance`: Waktu perkuliahan (1=Kelas Pagi/Reguler, 0=Kelas Sore-Malam/Karyawan).
11. `Previous_qualification`: Jenjang pendidikan sebelum masuk kuliah (SMA, SMK, Diploma).
12. `Previous_qualification_grade`: Nilai ijazah pendidikan sebelumnya (skala 0 - 200).
13. `Admission_grade`: Nilai tes seleksi masuk universitas (skala 0 - 200).

### C. Kelompok Sosio-Ekonomi & Finansial (Sangat Penting!)
14. `Mothers_qualification`: Pendidikan terakhir ibu (SD, SMP, SMA, Sarjana, dll.).
15. `Fathers_qualification`: Pendidikan terakhir ayah (SD, SMP, SMA, Sarjana, dll.).
16. `Mothers_occupation`: Jenis pekerjaan ibu (karyawan, wiraswasta, buruh, ibu rumah tangga).
17. `Fathers_occupation`: Jenis pekerjaan ayah (karyawan swasta, buruh, PNS, dll.).
18. `Educational_special_needs`: Memiliki kebutuhan khusus / disabilitas (1=Ya, 0=Tidak).
19. `Debtor`: Apakah mahasiswa punya riwayat tunggakan pinjaman di kampus (1=Punya Tunggakan, 0=Lancar).
20. `Tuition_fees_up_to_date`: Status pelunasan biaya kuliah / SPP (1=Lunas/Tepat Waktu, 0=Menunggak SPP) ➔ *Salah satu indikator terkuat risiko putus studi!*
21. `Scholarship_holder`: Apakah mahasiswa menerima beasiswa (1=Penerima Beasiswa, 0=Biaya Sendiri).

### D. Kelompok Capaian Akademik Semester 1
22. `Curricular_units_1st_sem_credited`: Jumlah SKS semester 1 yang diakui dari transfer kredit.
23. `Curricular_units_1st_sem_enrolled`: Jumlah SKS semester 1 yang diambil di KRS (misal 20 SKS).
24. `Curricular_units_1st_sem_evaluations`: Frekuensi mahasiswa mengikuti ujian/asesmen di semester 1.
25. `Curricular_units_1st_sem_approved`: Jumlah SKS semester 1 yang **BERHASIL LULUS** (nilai >= batas minimal). Jika ambil 20 SKS tapi yang lulus hanya 2 SKS, risiko dropout sangat tinggi!
26. `Curricular_units_1st_sem_grade`: Nilai rata-rata semester 1 (skala 0 - 20).
27. `Curricular_units_1st_sem_without_evaluations`: Jumlah mata kuliah yang diambil tapi ditinggalkan tanpa mengikuti ujian.

### E. Kelompok Capaian Akademik Semester 2
28. `Curricular_units_2nd_sem_credited`: Jumlah SKS semester 2 yang diakui dari alih kredit.
29. `Curricular_units_2nd_sem_enrolled`: Jumlah SKS semester 2 yang dikontrak di KRS.
30. `Curricular_units_2nd_sem_evaluations`: Frekuensi mengikuti ujian di semester 2.
31. `Curricular_units_2nd_sem_approved`: Jumlah SKS semester 2 yang **BERHASIL LULUS**.
32. `Curricular_units_2nd_sem_grade`: Nilai rata-rata semester 2 (skala 0 - 20).
33. `Curricular_units_2nd_sem_without_evaluations`: Jumlah mata kuliah semester 2 yang ditinggalkan tanpa evaluasi.

### F. Kelompok Konteks Ekonomi Nasional (Faktor Eksternal)
34. `Unemployment_rate`: Tingkat pengangguran regional/nasional saat mahasiswa menempuh studi.
35. `Inflation_rate`: Tingkat inflasi ekonomi.
36. `GDP`: Pertumbuhan Produk Domestik Bruto (kondisi pertumbuhan ekonomi makro).

### G. Label Sasaran Prediksi (Target Ground Truth)
37. `Status`: Nasib akhir kelulusan mahasiswa:
    * `Graduate`: Berhasil lulus menyelesaikan studi (2.209 mahasiswa / ~49,9%).
    * `Dropout`: Putus studi / berhenti di tengah jalan (1.421 mahasiswa / ~32,1%).
    * `Enrolled`: Masih berstatus aktif berkuliah saat sensus data diambil (794 mahasiswa / ~18,0%).

```python
# -----------------------------------------------------------------------------
# 1. LOAD DATASET MENTAH DARI FOLDER data/raw/
# -----------------------------------------------------------------------------
import pandas as pd
import numpy as np
import os

data_path = '../data/raw/predict_students_dropout_and_academic_success.csv'
df = pd.read_csv(data_path, sep=';')

print("=== STATUS PEMUATAN DATASET ===")
print(f"File ditemukan di : {os.path.abspath(data_path)}")
print(f"Dimensi Data      : {df.shape[0]} baris x {df.shape[1]} kolom")
df.head()
```

```python
# -----------------------------------------------------------------------------
# 2. AUDIT NILAI HILANG (MISSING VALUE AUDIT)
# -----------------------------------------------------------------------------
missing_total = df.isnull().sum().sum()
print("=== HASIL AUDIT NILAI KOSONG ===")
if missing_total == 0:
    print("Hasil: Bersih! Tidak ada nilai kosong (missing values) pada dataset mentah.")
else:
    print(f"Ditemukan {missing_total} sel kosong pada kolom:")
    print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
# -----------------------------------------------------------------------------
# 3. ANALISIS KETIMPANGAN KELAS (CLASS IMBALANCE ANALYSIS)
# -----------------------------------------------------------------------------
target_counts = df['Status'].value_counts()
target_pct = (df['Status'].value_counts(normalize=True) * 100).round(2)

ringkasan_target = pd.DataFrame({
    'Kategori': target_counts.index,
    'Jumlah Mahasiswa': target_counts.values,
    'Persentase (%)': target_pct.values
})
print("=== DISTRIBUSI KELAS TARGET (STATUS MAHASISWA) ===")
display(ringkasan_target) if 'display' in globals() else print(ringkasan_target)

print("\n💡 Insight Skripsi:")
print("Proporsi Dropout (~32%) lebih sedikit dibanding Graduate (~50%).")
print("Inilah alasan kuat mengapa teknik BORDERLINE-SMOTE wajib diterapkan pada tahap persiapan data!")
```

```python
# -----------------------------------------------------------------------------
# 4. PERBANDINGAN FITUR KRITIS: SKS LULUS & KEUANGAN (SPP)
# -----------------------------------------------------------------------------
# Perbandingan SKS Lulus Semester 1 berdasarkan Status Kelulusan
sks_summary = df.groupby('Status')[['Curricular_units_1st_sem_approved', 'Curricular_units_2nd_sem_approved']].mean().round(2)
print("=== RATA-RATA SKS YANG BERHASIL DILULUSKAN ===")
print(sks_summary)

# Perbandingan Mahasiswa yang Menunggak SPP vs Status Kelulusan
# Tuition_fees_up_to_date: 1 = Lunas, 0 = Menunggak SPP
spp_crosstab = pd.crosstab(df['Status'], df['Tuition_fees_up_to_date'], normalize='index') * 100
spp_crosstab.columns = ['Menunggak SPP (%)', 'SPP Lunas (%)']
print("\n=== HUBUNGAN TUNGGAKAN SPP DENGAN KELULUSAN ===")
print(spp_crosstab.round(2))

print("\n💡 Temuan Kunci Lapangan:")
print("- Mahasiswa Dropout rata-rata hanya meluluskan ~2 SKS di Semester 1 dan ~1 SKS di Semester 2.")
print("- Mayoritas mahasiswa yang menunggak SPP berakhir pada status Dropout.")
```
